### Tuplas (Inmutable - Longitud Fija)

````python
# Crear la tupla a partir de una secuencia de valores entre parentesis separados por comas
estudiantes = ('Andres', 'Cristina','duvan','German')
print(estudiantes)
````

````python
#Revisar si un objeto pertenece a la tupla
print(f"duvan está en la tupla estudiantes? {'duvan' in estudiantes}")
print(f"El número 50 está en la tupla estudiantes? {50 in estudiantes}")
````

````python
#Acceder a los datos de la tupla
print(f"Primer elemento de la tupla: {estudiantes[0]}") #INDEXACION
print(f"Segundo elemento de la tupla: {estudiantes[1]}")
print(f"Tercer elemento de la tupla: {estudiantes[2]}")
````

````python
# Tuplas anidadas
tupla_candidato = ("Cristian", "cristian@gmail.com",("Analista de datos", ("Python",5)))
````

````python
tupla_candidato[2][1][1]
````

### Listas (Modificable - Longitud Variable)

````python
estudiantes = ['Mariana', 'Miguel', 'Nicol', 'Nina']
estudiantes
````

````python
## Funcion para insertar un elemento en una lista en la ultima posicion
estudiantes.append('Mariana')
estudiantes.append('Julian')
estudiantes.append('Pedro')
estudiantes.append('Tazmania')
````

````python
# Funcion para insertar un elemento en una lista en una posición específica
estudiantes.insert(2, 'Cristian')
````

````python
estudiantes
````

````python
"Mariana" in estudiantes
````

````python
## Funcion para encontrar la posición de un elemento en una lista
estudiantes.index("Cristian")
````

````python
#Las listas se pueden modificar
estudiantes[2] = 'David'
````

````python
estudiantes
````

````python
#Puedo remover objetos de la lista con pop()
estudiantes.pop() # Entre parentesis se especifica la posición del objeto que quiero remover
````

````python
# si no se especifica ningun valor se remueve el ultimo objeto de la lista
estudiantes.pop() 
````

````python
# Entre parentesis se especifica el objeto que quiero remover
estudiantes.remove("Mariana")
````

### Diccionarios (Llave: Valor)

````python
estudiantes =  {
    "15489842":
        {
            "Nombre": "Cristian",
            "Edad": 28,
            "Correo": "cristian@gmail.com",
            "Cargo": "Analista de datos",
            "Fecha Contratacion": "2026-11-24"
        },
    "15489843":
        {
            "Nombre": "Jazmin",
            "Edad": 30,
            "Correo": "jazmin@gmail.com",
            "Cargo": "Cientifica de datos"
        },
}
````

````python
estudiantes.keys()
````

````python
busqueda = "cristian@gmail.com"
for estudiante_id in estudiantes.keys():
    estudiante = estudiantes[estudiante_id]
    
    if estudiante["Correo"] == busqueda:
        print("Encontramos al estudiante con el correo cristian@gmail.com")
        print(estudiante['Nombre'])
        print(estudiante['Edad'])
        print(estudiante['Cargo'])
````

````python
estudiantes["15489842"]['Nivel Estudio'] = "Maestria"
````

````python
estudiantes
````

````python
estudiantes["165465169"] = {
    "Nombre": "Juan",
    "Edad": 22,
    "Correo": "juan@gmail.com",
    "Cargo": "Ingeniero de datos"
}
````

````python
estudiantes['165465169']['Edad']
````

````python
estudiantes.pop('165465169')
````

````python
# Para agregar nuevas llaves al diccionario, declaro la nueva llave entre parentesis cuadrados y le asigno un nuevo valor
estudiantes[165465165]["edad"] = 31
estudiantes[165465165]["correo"] = "cristian.garcia@example.com"
estudiantes[165465165]
````

````python
# Puedo borrar datos del diccionario usando la sentencia 'del' delete
del estudiantes[165465169]
estudiantes
````

````python
# Tambien puedo borrar datos del diccionario usando la funcion pop() especificando la llave que quiero eliminar
estudiantes.pop(165465165)
estudiantes
````
